```python
# Implementasi siklus fetch-cycle

mem = [0] * 12                                                      # Buat 12 lokasi memori
pc = 0                                                              # Inisialisasi program counter (pc) dengan 0

mem[0] = 0b000100001100                                             # Instruksi 1: load r0 with 12

mem[1] = 0b001000000111                                             # Instruksi 2: add mem[7] to r0

mem[2] = 0b111100000000                                             # Instruksi 3: stop

mem[7] = 10                                                          # Data inisial di lokasi 7 adalah 8
def fetch(memory):                                                  # Function for fetch phase
    global pc                                                       # Buat pc sebagai variabel global untuk bisa diakses loop di bawah
    ir = memory[pc]                                                 # Baca instruksi dan pindahkan ke Instruction Register (IR)
    pc = pc + 1                                                     # Increment pc untuk siklus berikutnya
    # >> mengekstrak bit dari 0 ke 4 dengan menggeser 8 ke kanan
    # & 0xFF mask opcode ke 8-bit (0xFF = 255 dalam hex)
    return(ir >> 8, ir & 0xFF)                                      # Returns opCode dan operand, opcode diperoleh dengan bitwise right shift operator, operand dengan bitwise OR
run = 1                                                             # run = 1 untuk continue
while run == 1:                                                     # REPEAT: The program execution loop
    opCode, address = fetch(mem)                                    # Panggil fetch untuk perform fetch phase
    if opCode == 0b1111: run = 0                                    # Jalankan phase untuk stop (set run to 0 on stop)
    elif opCode == 0b0001:                                          # Jalankan phase untuk load number
        r0 = address                                                # Load r0 with contents of address field
    elif opCode == 0b0010:                                          # Jalankan phase untuk add
        mar = address                                               # Copy address dari opCode ke MAR
        mbr = mem[mar]                                              # Baca angka untuk di add
        r0 = mbr + r0                                               # Lakukan addition
    print('pc = ',pc - 1, 'opCode =', opCode, 'Register r0 =',r0)   # Print pc – 1 karena pc telah ditambahkan 1
```
