```python
import torch
import torch.nn as nn
import torch.optim as optim

# 1. Definisikan/Inisialisasi Model
model = nn.Linear(1, 1)

# 2. Definisikan Loss Function dan Optimizer
loss_fn = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# 3. Siapkan Data Dummy (jika belum ada)
x = torch.randn(10, 1)
y_true = 2 * x + 1

# 4. Training Loop
epochs = 100

for epoch in range(epochs):
    # Forward Pass
    y_hat = model(x)

    # Calculate Loss
    loss = loss_fn(y_hat, y_true)

    # Reset Gradien, Backward, dan Update Bobot
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch % 10 == 0:
        print(f"epoch {epoch:02d}: Loss= {loss.item():.4f}")
```
